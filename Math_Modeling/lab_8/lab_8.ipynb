## 1. Исходная гидродинамическая модель

Исходная физическая система представляет собой классическую модель конвекции Рэлея-Бенара. Она описывает плоский горизонтальный слой вязкой жидкости, который равномерно подогревается снизу. 

Математически эта система описывается сложным набором дифференциальных уравнений в частных производных:
*   Уравнениями Навье-Стокса для движения вязкой жидкости.
*   Уравнением теплопроводности.
*   Уравнением неразрывности (в приближении Буссинеска, предполагающем постоянство плотности везде, кроме члена, описывающего подъемную силу Архимеда).

## 2. Вывод модели Лоренца

$$ \dot{x} = -\sigma x + \sigma y $$
$$ \dot{y} = rx - y - xz $$
$$ \dot{z} = xy - bz $$

Физический смысл переменных:
*   $x$ — пропорциональна интенсивности конвективного движения (скорости вращения конвективных валов).
*   $y$ — характеризует разность температур между восходящим и нисходящим потоками жидкости.
*   $z$ — описывает отклонение вертикального профиля температуры от линейного равновесного распределения.
*   $r$ — нормированное число Рэлея (пропорционально разности температур между нижней и верхней поверхностями слоя).

**1. Исходные дифференциальные уравнения (приближение Буссинеска)**
Запишем систему уравнений Навье-Стокса и теплопроводности для двумерной конвекции жидкости, подогреваемой снизу, через функцию тока $\psi(x,z,t)$ и отклонение температуры от линейного профиля $\theta(x,z,t)$:


$$\frac{\partial}{\partial t}\nabla^2\psi + \frac{\partial(\psi, \nabla^2\psi)}{\partial(x, z)} = \sigma \nabla^4\psi + \sigma \frac{\partial \theta}{\partial x}$$

$$\frac{\partial \theta}{\partial t} + \frac{\partial(\psi, \theta)}{\partial(x, z)} = r \frac{\partial \psi}{\partial x} + \nabla^2\theta$$


где $\sigma$ — число Прандтля, $r$ — нормированное число Рэлея, а конвективные нелинейные члены выражены через якобиан: $\frac{\partial(f, g)}{\partial(x, z)} = \frac{\partial f}{\partial x}\frac{\partial g}{\partial z} - \frac{\partial f}{\partial z}\frac{\partial g}{\partial x}$.

**2. Аппроксимация методом Галёркина**
Задаем краевые условия для свободных границ ($\psi = \nabla^2\psi = \theta = 0$ при $z=0, \pi$). Искомые функции раскладываем в двойные ряды Фурье и усекаем до трех основных пространственных гармоник. Для удобства сразу введем лоренцевские нормировочные коэффициенты:


$$\psi(x,z,t) = \frac{\sqrt{2}(1+a^2)}{a} X(t) \sin(ax)\sin(z)$$

$$\theta(x,z,t) = \frac{\sqrt{2}(1+a^2)^3}{a^2} Y(t) \cos(ax)\sin(z) - \frac{(1+a^2)^3}{a^2} Z(t) \sin(2z)$$


где $a$ — волновое число (геометрический параметр конвективной ячейки вдоль оси $x$), а $X(t), Y(t), Z(t)$ — неизвестные функции амплитуд, зависящие только от времени.

**3. Применение операторов и тригонометрические преобразования**
Лапласианы для выбранных базисных функций дают:


$$\nabla^2 (\sin(ax)\sin(z)) = -(a^2+1)\sin(ax)\sin(z)$$

$$\nabla^2 (\cos(ax)\sin(z)) = -(a^2+1)\cos(ax)\sin(z)$$

$$\nabla^2 (\sin(2z)) = -4\sin(2z)$$

Раскроем нелинейный член конвективного теплопереноса $\frac{\partial(\psi, \theta)}{\partial(x,z)}$. Вычисляем частные производные:


$$\frac{\partial \psi}{\partial x} \sim \cos(ax)\sin(z), \quad \frac{\partial \psi}{\partial z} \sim \sin(ax)\cos(z)$$

$$\frac{\partial \theta}{\partial x} \sim \sin(ax)\sin(z), \quad \frac{\partial \theta}{\partial z} \sim \cos(ax)\cos(z) - \cos(2z)$$


Перемножение этих градиентов дает взаимодействие пространственных мод:

1. Произведение $\sin(z)\cos(z) = \frac{1}{2}\sin(2z)$ генерирует моду для переменной $Z(t)$.
2. Взаимодействие поля скоростей с $\cos(2z)$ создает гармонику $\cos(ax)\sin(z)$, питающую переменную $Y(t)$.
3. Слагаемые, содержащие $\sin(2ax)$, при усечении системы отбрасываются (метод Галеркина подразумевает ортогональное проектирование на исходный базис).

**4. Замыкание динамической системы**
Подставляем вычисленные выражения обратно в систему из шага 1. Сравниваем коэффициенты при одинаковых ортогональных функциях $\sin(ax)\sin(z)$, $\cos(ax)\sin(z)$ и $\sin(2z)$ и переходим к безразмерному времени $\tau = (1+a^2)t$. Получаем искомую динамическую систему Лоренца:

*Из уравнения динамики (при гармонике $\sin(ax)\sin(z)$):*


$$\dot{X} = -\sigma X + \sigma Y$$


*Из уравнения теплопроводности (при гармонике $\cos(ax)\sin(z)$):*


$$\dot{Y} = r X - Y - X Z$$


*Из уравнения теплопроводности (при гармонике $\sin(2z)$):*


$$\dot{Z} = X Y - b Z$$


где параметр $b$ возникает из действия оператора Лапласа на моду $\sin(2z)$ и равен $b = \frac{4}{1+a^2}$. Приняв формальную замену амплитуд $X, Y, Z \to x, y, z$, получаем классический вид модели.

## 3. Особые точки (стационарные решения)

Для нахождения особых точек системы при заданных параметрах $\sigma = 9$ и $b = 3$, приравняем правые части уравнений к нулю:

$$ -9x + 9y = 0 $$
$$ rx - y - xz = 0 $$
$$ xy - 3z = 0 $$

Из первого уравнения следует:
$$ x = y $$

Подставим это во второе и третье уравнения:
$$ x(r - 1 - z) = 0 $$
$$ x^2 - 3z = 0 \implies z = \frac{x^2}{3} $$

Из уравнения $x(r - 1 - z) = 0$ возникают два случая:
1.  **Случай $x = 0$:**
    Тогда $y = 0$ и $z = 0$. Мы получаем первую особую точку — начало координат $O(0, 0, 0)$. Она физически означает состояние покоя жидкости (теплопередача идет только за счет теплопроводности). Эта точка существует при любых значениях $r$.

2.  **Случай $x \neq 0$:**
    Тогда $r - 1 - z = 0 \implies z = r - 1$.
    Подставляем найденное $z$ в третье уравнение:
    $$ x^2 = 3(r - 1) \implies x = \pm\sqrt{3(r - 1)} $$
    Так как $x = y$, мы получаем еще две особые точки:
    $$ C_1\left(\sqrt{3(r - 1)}, \sqrt{3(r - 1)}, r - 1\right) $$
    $$ C_2\left(-\sqrt{3(r - 1)}, -\sqrt{3(r - 1)}, r - 1\right) $$
    Эти точки соответствуют режиму стационарной конвекции (валы вращаются в левую или правую сторону). Поскольку подкоренное выражение должно быть неотрицательным, точки $C_1$ и $C_2$ существуют только при $r \ge 1$.

## 4. Устойчивость особых точек

Составим матрицу Якоби для нашей системы:
$$ J = \begin{pmatrix} -9 & 9 & 0 \\ r-z & -1 & -x \\ y & x & -3 \end{pmatrix} $$

### Устойчивость точки $O(0, 0, 0)$
Подставим координаты точки $O$ в матрицу Якоби:
$$ J_0 = \begin{pmatrix} -9 & 9 & 0 \\ r & -1 & 0 \\ 0 & 0 & -3 \end{pmatrix} $$

Собственные значения находим из характеристического уравнения $\det(J_0 - \lambda E) = 0$. Очевидно, что $\lambda_3 = -3$. Для оставшегося блока $2 \times 2$ уравнение имеет вид:
$$ (-9 - \lambda)(-1 - \lambda) - 9r = 0 $$
$$ \lambda^2 + 10\lambda - 9(r - 1) = 0 $$

Корни уравнения:
$$ \lambda_{1,2} = -5 \pm \sqrt{25 + 9(r - 1)} = -5 \pm \sqrt{16 + 9r} $$

*   При **$r < 1$** все три собственных значения отрицательны. Точка $O$ является **устойчивым узлом**.
*   При **$r > 1$** корень $\lambda_1$ становится положительным. Точка $O$ теряет устойчивость и становится **седлом**.

### Устойчивость точек $C_1$ и $C_2$ (при $r \ge 1$)
Характеристическое уравнение матрицы Якоби после подстановки координат точек $C_1$ или $C_2$ приводится к виду:
$$ \lambda^3 + 13\lambda^2 + 3(r + 9)\lambda + 54(r - 1) = 0 $$

Для анализа устойчивости применим критерий Рауса-Гурвица для кубического уравнения $\lambda^3 + A\lambda^2 + B\lambda + C = 0$.
Условия устойчивости: $A > 0$, $C > 0$, $AB > C$.

В нашем случае:
$$ A = 13 > 0 $$
$$ C = 54(r - 1) > 0 \quad (\text{так как } r > 1) $$
$$ B = 3(r + 9) $$

Проверим главное условие $AB > C$:
$$ 13 \cdot 3(r + 9) > 54(r - 1) $$
$$ 39(r + 9) > 54r - 54 $$
$$ 39r + 351 > 54r - 54 $$
$$ 405 > 15r \implies r < 27 $$

**Вывод:** Фокусы $C_1$ и $C_2$ устойчивы при $1 \le r < 27$.

## 5. Замкнутые инвариантные траектории и появление аттрактора

При прохождении параметра через критическое значение **$r = 27$** происходит субкритическая бифуркация Андронова-Хопфа. 
Устойчивые фокусы $C_1$ и $C_2$ теряют устойчивость, и вокруг них рождаются неустойчивые предельные циклы (замкнутые инвариантные траектории). Эти циклы становятся границами, отталкивающими фазовые траектории, что приводит к сложному поведению системы.

### Свойства для появления аттрактора:
1.  **Диссипативность (сжатие фазового объема):**
    Вычислим дивергенцию векторного поля системы:
    $$ \text{div} \vec{V} = \frac{\partial \dot{x}}{\partial x} + \frac{\partial \dot{y}}{\partial y} + \frac{\partial \dot{z}}{\partial z} = -\sigma - 1 - b = -9 - 1 - 3 = -13 $$
    Так как $\text{div} \vec{V} < 0$, фазовый объем экспоненциально сжимается. Все траектории притягиваются к множеству нулевого объема — аттрактору.
2.  **Размерность:** Система должна состоять не менее чем из трех нелинейных дифференциальных уравнений (выполняется).
3.  **Чувствительность к начальным условиям:** В системе должен присутствовать хотя бы один положительный показатель Ляпунова, что вызывает экспоненциальное расхождение близких траекторий.

## 6. Фрактальная размерность

Фрактальная размерность аттрактора оценивается по формуле Каплана-Йорке:
$$ D_L = j + \frac{\sum_{i=1}^j \lambda_i}{|\lambda_{j+1}|} $$
где $\lambda_i$ — спектр показателей Ляпунова, упорядоченный по убыванию ($\lambda_1 \ge \lambda_2 \ge \lambda_3$), а $j$ — максимальное число, при котором сумма показателей еще положительна.

Сумма всех трех показателей Ляпунова равна дивергенции векторного поля, то есть $-13$. 
В режиме развитого странного аттрактора (например, при $r = 28$):
*   $\lambda_1 \approx 0.9$ (отвечает за разбегание траекторий, хаос).
*   $\lambda_2 = 0$ (отвечает за сдвиг вдоль самой траектории).
*   $\lambda_3 \approx -13.9$ (обеспечивает общее сжатие фазового объема).

Подставляем значения:
$$ D_L = 2 + \frac{\lambda_1 + \lambda_2}{|\lambda_3|} = 2 + \frac{0.9 + 0}{|-13.9|} \approx 2.06 $$

Дробная размерность, превышающая 2, указывает на сложную, бесконечно вложенную слоистую структуру аттрактора (фрактал).

## 7. Выбор 5 значений параметра r для моделирования

Для численного исследования выбраны следующие репрезентативные значения:

1.  **$r = 0.5$** — Режим теплопроводности. Все траектории сходятся к единственной устойчивой точке — началу координат $O(0,0,0)$.
2.  **$r = 15$** — Стационарная конвекция. Траектории притягиваются к одному из устойчивых фокусов $C_1$ или $C_2$.
3.  **$r = 24$** — Переходный хаос. Наблюдается длительное хаотическое блуждание траектории между центрами, но в итоге она все равно стабилизируется в одной из точек $C_1$ или $C_2$.
4.  **$r = 27$** — Точка бифуркации Хопфа. Граница потери устойчивости стационарных конвективных режимов.
5.  **$r = 28$** — Развитый хаос. Формируется классический странный аттрактор Лоренца ("бабочка"). Траектории совершают апериодические колебания, бесконечно перескакивая из окрестности одной особой точки в окрестность другой.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

sigma = 9
b = 3
r_values = [0.5, 15, 24, 27, 28]
initial_state = [1.0, 1.0, 1.0]
t_span = (0, 50)
t_eval = np.linspace(t_span[0], t_span[1], 15000)

def lorenz_system(t, state, r_val):
    x, y, z = state
    dxdt = -sigma * x + sigma * y
    dydt = r_val * x - y - x * z
    dzdt = x * y - b * z
    return [dxdt, dydt, dzdt]

fig = plt.figure(figsize=(20, 12))

for i, r in enumerate(r_values):
    sol = solve_ivp(lorenz_system, t_span, initial_state, args=(r,), t_eval=t_eval, rtol=1e-8, atol=1e-8)
    
    ax = fig.add_subplot(2, 3, i + 1, projection='3d')
    ax.plot(sol.y[0], sol.y[1], sol.y[2], color='#007acc', lw=0.6, alpha=0.8)
    
    if r >= 1:
        cp_val = np.sqrt(b * (r - 1))
        z_val = r - 1
        ax.scatter([0, cp_val, -cp_val], [0, cp_val, -cp_val], [0, z_val, z_val], color='#ff2a2a', s=60, zorder=5)
    else:
        ax.scatter([0], [0], [0], color='#ff2a2a', s=60, zorder=5)
        
    ax.set_title(f"r = {r}", fontsize=16, fontweight='bold', pad=15)
    ax.set_xlabel("Ось X", fontsize=12)
    ax.set_ylabel("Ось Y", fontsize=12)
    ax.set_zlabel("Ось Z", fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
